
# Linear Algebra Project on Student Performance Dataset

This notebook covers:
- Vector and Matrix Fundamentals
- Matrix Operations
- Eigenvalues & Eigenvectors
- LU Decomposition
- Singular Value Decomposition (SVD)
- Principal Component Analysis (PCA)
- Linear Discriminant Analysis (LDA)


```

import pandas as pd
import numpy as np

df = pd.read_csv('/mnt/data/student_performance_1000.csv')
df.head()
```


## Part A: Vector & Matrix Fundamentals
Represent student scores as vectors.


```

subjects = ['Mathematics', 'Physics', 'Chemistry', 'Computer_Science', 'English']

student1 = df.loc[0, subjects].values
student2 = df.loc[1, subjects].values

print("Student 1 Vector:", student1)
print("Student 2 Vector:", student2)
```

```

# Norm-1 and Norm-2
norm1 = np.linalg.norm(student1, ord=1)
norm2 = np.linalg.norm(student1, ord=2)

print("Norm-1:", norm1)
print("Norm-2:", norm2)
```

```

# Dot Product
dot_product = np.dot(student1, student2)
print("Dot Product:", dot_product)
```

```

# Cross Product (first 3 values)
cross_product = np.cross(student1[:3], student2[:3])
print("Cross Product:", cross_product)
```


## Part B: Matrix Operations


```

matrix = df[['Mathematics', 'Physics', 'Chemistry', 'Computer_Science', 'English']].head(5).values

print("Matrix:")
print(matrix)

print("\nTranspose:")
print(matrix.T)

determinant = np.linalg.det(matrix)
print("\nDeterminant:", determinant)
```


## Part C: Eigenvalues and Eigenvectors


```

cov_matrix = np.cov(df[['Mathematics', 'Physics', 'Chemistry', 'Computer_Science', 'English']].values.T)

eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

print("Eigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)
```


## Part D: LU Decomposition and SVD


```

from scipy.linalg import lu
from numpy.linalg import svd

P, L, U = lu(matrix)

print("L Matrix:")
print(L)

print("\nU Matrix:")
print(U)

U_svd, S_svd, Vt_svd = svd(matrix)

print("\nSingular Values:")
print(S_svd)
```


## Part E: PCA and LDA


```

from sklearn.decomposition import PCA
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

X = df[['Mathematics', 'Physics', 'Chemistry', 'Computer_Science', 'English']]

# PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

print("PCA Shape:", X_pca.shape)

# Create target for LDA
y = (df[Mathematics] > df[Mathematics].mean()).astype(int)

lda = LinearDiscriminantAnalysis(n_components=1)
X_lda = lda.fit_transform(X, y)

print("LDA Shape:", X_lda.shape)
```
